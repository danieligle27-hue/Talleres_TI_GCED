## Parte 1: Señales discretas y convenciones

### Contexto teórico

Una **señal discreta** es una secuencia de valores definida sobre un índice entero $n$. Una forma muy habitual (que usaremos en todo el taller) es:

$$x[n] = A \cdot \sin\left(2\pi \frac{f}{F_s} n\right) + C$$

donde:
- $A$ es la amplitud
- $f$ es la frecuencia de la señal (Hz)
- $F_s$ es la frecuencia de muestreo (Hz)
- $n$ es el índice discreto (muestras, no segundos)
- $C$ es una constante (offset o componente DC). En el **Bloque 1** se toma $C = 0$ (sinusoide sin offset); más adelante veremos el efecto de $C \neq 0$.

### Convención fundamental

| Símbolo | Significado | Tipo |
|---------|------------|------|
| `n` | Índice discreto | Enteros (0, 1, 2, ...) |
| `Fs` | Frecuencia de muestreo | Hz |
| `t = n/Fs` | Tiempo físico | Segundos |

**Error común:** Confundir `n` con `t`. El índice `n` **no tiene unidades de tiempo**.

### Bloque 1 · Primera sinusoide y parámetros base

---

#### 📝 Hipótesis previa (COMPLETA ANTES de ejecutar)

**Pregunta:** Si $F_s = 50$ Hz y $T = 2$ s, ¿cuántas muestras tendrá la señal?

*Tu predicción:* N = `_____100_______`

---

**Parámetros:** (en este bloque usamos $C = 0$, es decir, sin offset)
- Amplitud: $A = 1$
- Frecuencia de la señal: $f = 2$ Hz
- Frecuencia de muestreo: $F_s = 50$ Hz
- Duración: $T = 5$ s
- Constante: $C = 0$

**Tareas:**
1. Calcula el número de muestras $N$.
2. Crea el índice discreto $n = 0, 1, 2, \ldots, N-1$.
3. Calcula el instante de tiempo físico $t$.
4. Genera la señal $x[n]$ con la expresión anterior (con $C = 0$).

````python
# === PARÁMETROS BASE (usados en todo el taller) ===
A = 1
f = 2        # Hz
Fs = 50      # Hz
T = 5        # segundos

# === IMPLEMENTACIÓN ===
N = int(Fs * T)  # Número total de muestras  
t = np.linspace(0, T, N, endpoint=False)  # Vector de tiempo  
n = np.arange(N)
x = A * np.sin(2 * np.pi * f * n/Fs)
````

````python
# Ejecuta esta celda para validar tu implementación.
# === VALIDACIÓN ===
assert N == int(Fs * T), f"N debería ser {int(Fs*T)}, pero es {N}"
assert len(n) == N, "El índice n debe tener N elementos"
assert np.allclose(t, n / Fs), "t debe ser n/Fs"
assert len(x) == N, "La señal x debe tener N muestras"
assert np.isclose(np.max(np.abs(x)), A, atol=0.01), "Amplitud máxima debería ser A"

print(f"✓ N = {N} muestras, duración = {T} s, Fs = {Fs} Hz")
print(f"✓ Señal generada: amplitud máxima = {np.max(np.abs(x)):.3f}")
````

## Parte 2: Medidas básicas — media, energía y potencia

### Contexto teórico

Para una señal discreta de $N$ muestras:

| Medida | Fórmula | Interpretación |
|--------|---------|----------------|
| **Media** | $\mu = \frac{1}{N}\sum_{n=0}^{N-1} x[n]$ | Valor promedio (componente DC) |
| **Energía** | $E = \sum_{n=0}^{N-1} \lvert x[n] \rvert^2$ | Acumulado total (crece con N) |
| **Potencia media** | $P = \frac{1}{N}\sum_{n=0}^{N-1} \lvert x[n] \rvert^2$ | Promedio de energía por muestra |

**Pregunta clave:** ¿Qué pasa con E y P al aumentar N para una señal periódica?

### Bloque 3 · Efecto del offset en la media

---

#### 📝 Hipótesis previa

**Pregunta:** Si en la expresión general tomas $C = 0.5$ (en lugar de $C = 0$), es decir $x_{\text{offset}}[n] = x[n] + C$:
- Media de $x[n]$ original (con $C = 0$): esperada = 0
- Media de $x_{\text{offset}}[n] = x[n] + C$ con $C = 0.5$: esperada = 0.5

*Tu predicción:* `La señal original tiene una media de cero, y si se le suma una constante, tendrá de media esa constante`

---

````python
# === IMPLEMENTACIÓN ===


# TODO: Crea la señal con offset, x_offset
offset = 0.5
x_offset = x + offset
media_x = np.mean(x)
media_x_offset = np.mean(x_offset)
# TODO: Calcula las medias, media_x y media_x_offset

print(f"Media de x[n]: {media_x:.4f}")
print(f"Media de x_offset[n]: {media_x_offset:.4f}")
print(f"Diferencia: {media_x_offset - media_x:.4f}")
````

````python
# Ejecuta esta celda para validar tu implementación.
# === VALIDACIÓN ===
assert np.isclose(media_x_offset - media_x, offset, atol=0.01), \
    f"La diferencia de medias debería ser {offset}"
print("✓ El offset afecta a la media como se esperaba")
````

### Bloque 4 · Cálculo de energía y potencia

---

#### 📝 Hipótesis previa

**Pregunta:** Si la duración de la señal se duplica (de 2s a 4s), ¿cómo cambian:
- La energía E: ¿se duplica / se mantiene / otra cosa?
- La potencia P: ¿se duplica / se mantiene / otra cosa?

*Tu predicción:* `La energía se va a duplicar al duplicarse su duración, mientras que la potencia se va a mantener, al tener que dividir al final del sumatorio entre la longitud de la señal`

---

````python
# === IMPLEMENTACIÓN ===
# TODO: Implementa funciones de media, energía y potencia

def calcular_media(signal):
    suma = 0
    N = len(signal)
    for i in signal:
        suma += i
    media = suma/N
    """Calcula la media de una señal, sin usar la función np.mean"""
    return media  # Completa

def calcular_energia(signal):
    """Calcula la energía total de una señal."""
    suma_energia = 0
    N = len(signal)
    for i in signal:
        suma_energia += i**2
    energia = suma_energia
    return energia  # Completa

def calcular_potencia(signal):
    """Calcula la potencia media de una señal."""
    N = len(signal)
    suma_energia = 0
    for i in signal:
        suma_energia += i**2
    energia = suma_energia
    return energia  # Completa

# Calcula para ambas señales
mu_x, E_x, P_x = calcular_media(x), calcular_energia(x), calcular_potencia(x)
mu_off, E_off, P_off = calcular_media(x_offset), calcular_energia(x_offset), calcular_potencia(x_offset)

# Muestra resultados en formato tabla
print(f"{'Medida':<15} {'x[n]':>12} {'x_offset[n]':>12}")
print("-" * 40)
print(f"{'Media':<15} {mu_x:>12.4f} {mu_off:>12.4f}")
print(f"{'Energía':<15} {E_x:>12.2f} {E_off:>12.2f}")
print(f"{'Potencia':<15} {P_x:>12.4f} {P_off:>12.4f}")
````

````python
# Ejecuta esta celda para validar tu implementación.
# === VALIDACIÓN ===
assert P_x > 0, "La potencia debe ser positiva"
assert E_x > 0, "La energía debe ser positiva"
assert P_off > P_x, "La potencia con offset debería ser mayor"
print("\n✓ Medidas calculadas correctamente")
````

## Parte 3: Señales compuestas y operaciones básicas

### Contexto teórico

Una señal compuesta combina múltiples componentes:

$$x[n] = A_1 \sin\left(2\pi \frac{f_1}{F_s} n\right) + A_2 \cos\left(2\pi \frac{f_2}{F_s} n\right) + C$$


### Bloque 5 · Señal con dos componentes frecuenciales

---

**Parámetros:**
- $A_1 = 1$
- $f_1 = 2$ Hz con amplitud 0.7
- $A_2 = 0.5$
- $f_2 = 7$ Hz con amplitud 0.3
- Offset $C = 0.2$
- Misma duración y frecuencia de muestreo que en los bloques anteriores

---

````python
# === IMPLEMENTACIÓN ===

# TODO: Construye la señal compuesta, x_comp
A1 = 1
f1 = 2    # Hz
A2 = 0.5
f2 = 7    # Hz
C = 0.2   # Constante DC
x_comp = A1 * np.sin(2 * np.pi * f1 * t) + A2 * np.sin(2 * np.pi * f2 * t) + C
# TODO: Representa un zoom de los primeros 60 puntos
n_zoom = n[:60]
t_zoom = t[:60]
x_zoom = x_comp[:60]

plt.figure(figsize=(10, 3))
# Removed `use_line_collection` for compatibility with some matplotlib versions
plt.stem(t_zoom, x_zoom, basefmt=' ')
plt.title('Zoom: primeros 60 puntos de x_comp')
plt.xlabel('t (s)')
plt.ylabel('x[n]')
plt.tight_layout()
plt.show()


# Verifica la media
media_comp = np.mean(x_comp)
print(f"Media de x_comp: {media_comp:.4f} (esperada ≈ {C})")
````

````python
# Ejecuta esta celda para validar tu implementación.
# === VALIDACIÓN ===
assert x_comp is not None and len(x_comp) == N
assert abs(media_comp - C) < 0.05, f"Media ({media_comp:.3f}) debería estar cerca de {C}"
print("✓ Señal compuesta generada correctamente")
````

### Bloque 6 · Segmentación: extraer 1 período y añadir ceros

---

#### Objetivo del bloque

Preparar una señal **finita** a partir de la sinusoide del Bloque 1 para poder aplicar después un retardo **sin confusiones**.

La idea es:
- Tomar un segmento que corresponda con **un período** de la señal original $x[n]$.
- Construir una nueva señal de longitud fija con **ese segmento** y, a continuación, **una secuencia de ceros** a la derecha de la misma longitud que el segmento extraido.

---

#### Definición

Si $N_p$ es el número de muestras por período:

$$x_{\text{finita}}[n] = \begin{cases}
 x[n] & 0 \le n < N_p\\
 0 & N_p \le n < 2N_p
\end{cases}$$

---

#### 📝 Hipótesis previa

**Pregunta:** ¿Qué longitud tendrá $x_{\text{finita}}[n]$ y qué valores tendrá en su segunda mitad? ____

---

````python
# === IMPLEMENTACIÓN ===
# Vamos a extraer 1 período de la sinusoide del Bloque 1 y añadir otro período de ceros.

# Número de muestras por período
Np = int(round(Fs / f))  # muestras por período (Fs/f)

# TODO: Extrae un período de x, empezando en n=0. Rellena con ceros hasta llegar a 2Np, obteniendo, x_finita
period = x[:Np]
x_finita = np.concatenate([period, np.zeros(Np)])

# Índice para representar esta señal (longitud 2Np)
n_pad = np.arange(len(x_finita))

# Visualización
plt.figure(figsize=(12, 3))
plt.stem(n_pad, x_finita, basefmt=' ')
plt.title('Señal de prueba: 1 período + 1 período de ceros (x_pad)')
plt.xlabel('n')
plt.tight_layout()
plt.show()
````

### Bloque 7 · Operaciones: desplazamiento (shift) y escalado

---

#### Objetivo del bloque

Aplicar un **desplazamiento** y **escalado** sobre una señal finita `x_finita`.

- Deslazamiento, será un retardo de $n_0=10$ muestras, de forma que $x_{retardada}[n]=x_{finita}[n-n_0]$
  - Vamos a considerar que la señal $x_{retardada}$ sea de la misma longitud que $x_{finita}$.
- Escalado, consistirá en multiplicar la amplitud de la señal por $\alpha=2$, de forma que $x_{escalada}[n]=\alpha x_{finita}[n]$

---

#### 📝 Hipótesis previa

- Las primeras $n_0$ muestras tras el retardo serán ____
- ¿Cómo variará la potencia al desplazar la señal? ____
- ¿Qué factor esperas en la potencia al escalar por $\alpha$? ____

---

````python
# === IMPLEMENTACIÓN ===
# Generar la señal x_retardada a partir de x_finita construida en el Bloque anterior.

n0 = 10
alpha = 2.0

# TODO: Crea la señal x_retardada a partir de x_finita
x_retardada = np.zeros_like(x_finita)
if n0 < len(x_finita):
    x_retardada[n0:] = x_finita[:-n0]


# TODO: Crea la señala x_escalada a partir de x_finita
x_escalada = alpha * x_finita
# Potencia de las señales
P_x_finita = np.mean(x_finita**2)
P_x_retardada = np.mean(x_retardada**2)
P_x_escalada = np.mean(x_escalada**2)
print(f"Potencia de x_finita: {P_x_finita:.4f}")
print(f"Potencia de x_retardada: {P_x_retardada:.4f}")
print(f"Potencia de x_escalada: {P_x_escalada:.4f}")

# Visualización
fig, axes = plt.subplots(1, 3, figsize=(14, 3))

axes[0].stem(np.arange(len(x_finita)), x_finita, basefmt=' ')
axes[0].set_title('Original (x_finita)')
axes[0].set_xlabel('n')

axes[1].stem(np.arange(len(x_finita)), x_retardada, basefmt=' ', linefmt='r-', markerfmt='ro')
axes[1].set_title(f'Retardo n0={n0}')
axes[1].set_xlabel('n')

axes[2].stem(np.arange(len(x_finita)), x_escalada, basefmt=' ', linefmt='g-', markerfmt='go')
axes[2].set_title(f'Escalado α={alpha}')
axes[2].set_xlabel('n')

plt.tight_layout()
plt.show()
````

````python
# Ejecuta esta celda para validar tu implementación.
# === VALIDACIÓN ===

# Usamos los nombres de variables x_finita, x_retardada y x_escalada 
# para mantener coherencia con las señales definidas antes.

# Longitud fija
assert len(x_retardada) == len(x_finita)
assert len(x_escalada) == len(x_finita)

# Primeras n0 muestras del retardo: ceros
assert np.allclose(x_retardada[:n0], 0)

# Potencias (usamos la definición empírica P = (1/N) sum x^2)
P_x = np.mean(x_finita**2)
P_delay = np.mean(x_retardada**2)
P_scaled = np.mean(x_escalada**2)

# Si n0 < N, lo que se recorta por la derecha son ceros -> la potencia se conserva
assert np.isclose(P_delay, P_x, atol=1e-12)

# Escalado: P(alpha x) = alpha^2 P(x)
assert np.isclose(P_scaled, (alpha**2) * P_x, atol=1e-12)

print("✓ Retardo y escalado OK")
print(f"  P(x_finita)     = {P_x:.6f}")
print(f"  P(retardada)    = {P_delay:.6f} (debe coincidir)")
print(f"  P(escalada)     = {P_scaled:.6f} (debe ser α^2·P)")
````

### ✍️ Explicación (OBLIGATORIA)

**1. Un retardo de 10 muestras a $F_s = 50$ Hz, ¿cuántos milisegundos representa?**




**2. Si escalas por α=2, ¿cómo cambia la potencia? ¿Por qué?**


```
El cambio es de α^2*P. La potencia se queda como un sumatorio de 1/N*∑(α*x[n])^2. Alpha al ser una constante puede salir fuera del sumatorio, quedando α^2*P.
```

**3. ¿Por qué la energía del segmento es menor que la de la señal completa, pero la potencia puede ser similar?**


```
La energía es una magnitud acumulativa, por lo que al reducir la longitud de la señal mediante, esta se ve mermada.
Por otro lado, la potencia es un valor medio de la energía en un período, por lo que si el segmento escogido es lo suficientemente representativo, tendra un valor parecido al de la señal original.
```
